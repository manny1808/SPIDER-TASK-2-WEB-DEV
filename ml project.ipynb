```python
class Backtester():
    def __init__(self, symbol, SMA_S, SMA_L, start, end):
        self.symbol = symbol
        self.SMA_S = SMA_S
        self.SMA_L = SMA_L
        self.start = start
        self.end = end
        self.results = None
        self.get_data()

    def get_data(self):
        df = yf.download(self.symbol, start = self.start, end = self.end, auto_adjust = True)
        df.to_csv("data.csv")
        data = pd.read_csv("data.csv", header = [0, 1], index_col = [0], parse_dates = [0])
        data["returns"] = np.log(df.Close.div(df.Close.shift(1)))
        data["SMA_S"] = data.Close.rolling(self.SMA_S).mean()
        data["SMA_L"] = data.Close.rolling(self.SMA_L).mean()
        data.dropna(inplace = True)
        self.data2 = data

        return data

    def test_results(self):
        data = self.data2.copy().dropna()
        data["position_1"] = np.where(data["SMA_S"] > data["SMA_L"], 1, -1)
        data["strategy_1"] = data["returns"] * data["position_1"]
        data["position_2"] = np.where(data["SMA_S"] > data["SMA_L"], 1, 0)
        data["strategy_2"] = data["returns"] * data["position_2"]
        data.dropna(inplace = True)
        data["returns_bh"] = data["returns"].cumsum().apply(np.exp)
        data["returns_strategy_1"] = data["strategy_1"].cumsum().apply(np.exp)
        data["returns_strategy_2"] = data["strategy_2"].cumsum().apply(np.exp)
        self.results = data

        return data["returns_strategy_1"].iloc[-1] - data["returns_bh"].iloc[-1] , data["returns_strategy_2"].iloc[-1] - data["returns_bh"].iloc[-1]

    def plot_results(self):
        if self.results is None:
            print("RUN THE TEST FIRST")

        else:
            self.results[["returns_bh", "returns_strategy_1", "returns_strategy_2"]].plot(figsize = (10, 8), title = "COMPARISON")
            plt.show()
```

```python
tester = Backtester("MSFT", 50, 100, "2000-01-01", "2020-01-01")
```

```python
tester.test_results()
```

```python
tester.plot_results()
```
